# Análisis de Datos + Machine Learning - Sara Bakery
**Objetivo**: Predecir **cantidad_vendida** para optimizar inventario

**Dataset**: Ventas reales Sara Bakery (Enero-Marzo 2026)
**Variables predictoras**: sabor, precio_unitario, dia_semana
**Target**: cantidad_vendida

**Herramientas**: Pandas, Seaborn, Scikit-Learn

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score
import warnings
warnings.filterwarnings('ignore')

# DATASET SARA BAKERY
np.random.seed(42)
fechas = pd.date_range(start='2026-01-01', end='2026-03-31', freq='D')
sabores = ['chocolate', 'cookies_and_cream', 'red_velvet', 'vegan', 'sin_gluten']
dias_semana = ['lunes','martes','miércoles','jueves','viernes','sábado','domingo']

data = []
for fecha in fechas:
    for _ in range(np.random.randint(3,8)):
        data.append({
            'fecha': fecha,
            'sabor': np.random.choice(sabores),
            'cantidad_vendida': np.random.poisson(25) + np.random.randint(0,50),
            'precio_unitario': np.round(np.random.uniform(2.0, 3.5), 2),
            'dia_semana': fecha.day_name()
        })

df = pd.DataFrame(data)
print(f"✅ Dataset creado: {df.shape[0]} filas")
df.head()
```

```python
# Configurar gráficos bonitos
plt.style.use('default')
sns.set_palette("Set2")
fig, axes = plt.subplots(2, 2, figsize=(16,12))

# Gráfico 1: Ventas por sabor
sns.countplot(data=df, x='sabor', ax=axes[0,0])
axes[0,0].set_title('🧁 Distribución Ventas por Sabor', fontweight='bold')
axes[0,0].tick_params(axis='x', rotation=45)

# Gráfico 2: Boxplot por día semana
sns.boxplot(data=df, x='dia_semana', y='cantidad_vendida', ax=axes[0,1])
axes[0,1].set_title('📅 Ventas por Día de la Semana', fontweight='bold')
axes[0,1].tick_params(axis='x', rotation=45)

# Gráfico 3: Scatter precio vs cantidad
sns.scatterplot(data=df, x='precio_unitario', y='cantidad_vendida',
                hue='sabor', size='cantidad_vendida', ax=axes[1,0], alpha=0.6)
axes[1,0].set_title(' Precio vs Cantidad Vendida', fontweight='bold')

# Gráfico 4: Ingresos por sabor
df['ingresos'] = df['cantidad_vendida'] * df['precio_unitario']
top_sabores = df.groupby('sabor')['ingresos'].sum().sort_values(ascending=False)
sns.barplot(x=top_sabores.index, y=top_sabores.values, ax=axes[1,1])
axes[1,1].set_title(' Ingresos Totales por Sabor', fontweight='bold')
axes[1,1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# Estadísticas descriptivas
print(" Estadísticas Clave:")
print(df.groupby('sabor')['cantidad_vendida'].agg(['mean', 'sum']).round(1))
```
